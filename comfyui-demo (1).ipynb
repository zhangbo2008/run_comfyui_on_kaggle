```python
import subprocess, time, requests, os

os.chdir("/kaggle/working/")
!wget -q https://bin.equinox.io/c/bNyj1mQVY4c/ngrok-v3-stable-linux-amd64.tgz
!tar -xzf ngrok-v3-stable-linux-amd64.tgz
!chmod +x ngrok
# 设置你的ngrok token
NGROK_TOKEN = "2JLq5O6hEwAeTekrSjBSslTy5ML_2BevAgkzrcdee2JvSwwyZ"
subprocess.run(["./ngrok","config","add-authtoken", NGROK_TOKEN])

# 1.后台启动ComfyUI --listen 0.0.0.0
comfy_log = open("comfy.log","w")
proc_comfy = subprocess.Popen(
    ["python","main.py","--listen","0.0.0.0","--port","8188"],
    stdout=comfy_log, stderr=comfy_log
)
print(f"ComfyUI PID = {proc_comfy.pid}")

time.sleep(8)

#2.后台启动ngrok
ngrok_log = open("ngrok.log","w")
proc_ngrok = subprocess.Popen(
    ["./ngrok","http","8188"],
    stdout=ngrok_log, stderr=ngrok_log
)
print(f"ngrok PID = {proc_ngrok.pid}")

#等待建立隧道，最多25秒
wait_sec = 25
for i in range(wait_sec):
    try:
        r = requests.get("http://127.0.0.1:4040/api/tunnels",timeout=2)
        pub_url = r.json()["tunnels"][0]["public_url"]
        print("\n✅ ComfyUI访问链接：", pub_url)
        break
    except Exception:
        time.sleep(1)
else:
    print("❌超时，ngrok没有建立隧道，查看ngrok.log")
```
